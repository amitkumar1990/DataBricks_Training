```python
bronze_path = "/Volumes/Dev_Training/storage/Dev_Training/Bronze/users"
checkpoint_path = "/Volumes/Dev_Training/storage/Dev_Training/checkpoints/users"
schema_path = "/Volumes/Dev_Training/storage/Dev_Training/schema/users"
silver_table = "Dev_Training.bronze.users"

df = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "json")
    .option("cloudFiles.inferColumnTypes", "true")
    .option("cloudFiles.schemaLocation", schema_path)
    .load(bronze_path)
)

(
    df.writeStream
    .option("checkpointLocation", checkpoint_path)
    .trigger(availableNow=True)
    .toTable(silver_table)
)

```
